```python
dbutils.widgets.text("Name", "Dinesh")
name = dbutils.widgets.get("Name")
print(name)
```

```python
dbutils.widgets.dropdown("Courses", "Python", ["Python", "Java", "C++", "Scala"])
courses = dbutils.widgets.get("Courses")
print(courses)
```

```python
dbutils.widgets.combobox("Languages", "Telugu", ["Hindi", "Telugu", "English"])
lang = dbutils.widgets.get("Languages")
print(lang)
```

```python
dbutils.widgets.multiselect("Colors", "Red", ["Red", "Green", "Blue"])
colors = dbutils.widgets.get("Colors")
print(colors)
colors_list = colors.split(",")
print(colors_list)
```
