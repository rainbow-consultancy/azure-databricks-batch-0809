```python
cus_df = spark.table("samples.bakehouse.sales_customers")
trans_df = spark.table("samples.bakehouse.sales_transactions")
```

```python
cus_df.display()
```

```python
trans_df.display()
```

```python
trans_df.createOrReplaceTempView("trans")
```

```python
%sql 
select 
count(*) as total_rows, 
count(transactionID) as total_nonnull_transactions, 
count(distinct transactionID) as total_unique_records 
from trans
```

```python
cus_trans_df = (
    cus_df
    .join(
        trans_df, 
        on=["customerID"], 
        how="inner"
    )
    .select("customerID", "first_name", "email_address", "city", "transactionID", "totalPrice")
)
cus_trans_df.display()
```

```python
trans_df = trans_df.withColumnRenamed("customerID", "cus_id")
trans_df.display()
```

```python
cus_trans_df = (
    cus_df
    .join(
        trans_df, 
        on=cus_df.customerID == trans_df.cus_id, 
        how="inner"
    )
    .select("customerID", "first_name", "email_address", "city", "transactionID", "totalPrice")
)
cus_trans_df.display()
```

```python
# Sample data for demo
left_data = [
    (1, "Alice"),
    (2, "Bob"),
    (3, "Charlie"),
    (4, "David"),
    (5, "Eve")
]

right_data = [
    (2),
    (4),
    (5)
]

left_df = spark.createDataFrame(data=left_data, schema=["id", "name"])
right_df = spark.createDataFrame(data=right_data, schema=["id"])

left_df.createOrReplaceTempView("left_table")
right_df.createOrReplaceTempView("right_table")

display(left_df)
display(right_df)
```

```python
%sql
select *
from left_table
left semi join right_table
on left_table.id = right_table.id;
```

```python
%sql
select *
from left_table
left anti join right_table
on left_table.id = right_table.id;
```

```python
%sql
select l.*
from left_table l
left join right_table r
on l.id = r.id
where r.id is null;
```
