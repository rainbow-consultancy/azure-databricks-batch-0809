```python
from pyspark.sql import functions as F
```

```python
dbutils.fs.ls("abfss://adb-container@quantcloudadls.dfs.core.windows.net/")
```

```python
base_path = "abfss://adb-container@quantcloudadls.dfs.core.windows.net"
file_name = "videogames.csv"

video_games_df = spark.read.format("csv").option("header", True).option("inferSchema", True).load(f"{base_path}/{file_name}")
video_games_df.display()
```

```python
video_games_df.columns
```

```python
for col_name in video_games_df.columns:
    video_games_df = video_games_df.withColumnRenamed(col_name, col_name.lower())
```

```python
old_columns = video_games_df.columns
new_col_names_list = [col_name.lower() for col_name in old_columns]

video_games_df = (
    video_games_df
    .select(
        *[F.col(old_col_name).alias(new_col_name) 
        for new_col_name, old_col_name in zip(new_col_names_list, old_columns)]
    )  
)

video_games_df.display()
```

```python
video_games_df.write.format("parquet").save(f"{base_path}/processed/video_games")
```

```python
video_games_df = video_games_df.repartition(4)
video_games_df.write.format("parquet").mode("overwrite").save(f"{base_path}/processed/video_games")
```

```python
video_games_df.write.format("csv").mode("overwrite").save(f"{base_path}/processed/video_games_flat")
```
