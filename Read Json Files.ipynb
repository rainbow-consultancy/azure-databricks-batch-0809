```python
from pyspark.sql import functions as F
```

```python
base_path = "abfss://adb-container@quantcloudadls.dfs.core.windows.net"
folder_name = "Batch-0809-Final-Assignment"
```

```python
# Read Json File

emp_df = spark.read.format("json").option("multiline", "true").load(f"{base_path}/{folder_name}/employees_1MB.json")
display(emp_df)
```

```python
flattened_emp_df = (
    emp_df
    .select(
        F.col("employee.id").alias("employee_id"),
        F.col("employee.name").alias("employee_name"),
        F.col("employee.position").alias("position"),
        F.col("employee.department.id").alias("department_id"),
        F.col("employee.department.name").alias("department_name"),
        F.col("employee.department.manager.id").alias("manager_id"),
        F.col("employee.department.manager.name").alias("manager_name"),
        F.col("employee.department.manager.contact.email").alias("manager_email"),
        F.col("employee.department.manager.contact.phone").alias("manager_phone"),
        F.explode_outer("employee.projects").alias("projects")
    )
)

flattened_emp_df = (
    flattened_emp_df
    .select(
        flattened_emp_df["*"],
        F.col("projects.projectId").alias("project_id"),
        F.col("projects.projectName").alias("project_name"),
        F.col("projects.startDate").alias("project_start_date"),
        F.explode_outer("projects.tasks").alias("tasks")
    )
    .drop("projects")
)

display(flattened_emp_df)
```
