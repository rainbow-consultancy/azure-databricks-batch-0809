### Broadcast Join Stratagy

```python
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", "-1")
spark.conf.set("spark.databricks.adaptive.autoBroadcastJoinThreshold", "-1")
spark.conf.set("spark.sql.adaptive.enabled", "false")
```

```python
spark.conf.set("spark.sql.shuffle.partitions", "100")
```

```python
from pyspark.sql import functions as F
```

```python
users_df = spark.table("samples.wanderbricks.users")
bookings_df = spark.table("samples.wanderbricks.bookings")

result_df = (
    users_df.select("user_id", "name")
    .join(
        bookings_df,
        on=["user_id"],
        how="inner"

    )
)

result_df.display()
```

```python
users_df.count()
```

```python
# users_df.select(F.spark_partition_id()).distinct().count()
result_df.select(F.spark_partition_id()).distinct().count()
```

```python
users_df = spark.table("samples.wanderbricks.users")
bookings_df = spark.table("samples.wanderbricks.bookings")

result_df = (
    users_df.select("user_id", "name")
    .join(
        F.broadcast(bookings_df),
        on=["user_id"],
        how="inner"

    )
)

result_df.display()
```
