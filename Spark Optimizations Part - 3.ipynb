### repartition & coleasce

```python
# 128 MB is the default partition size
```

```python
# 1 GB = 1024 MB -> 1024/128 -> 8 partitions
# what if my df size is less than 128 MB -> 1 partition -> 8 partitions
```

```python
# repartition -> it can be used either to increase the no.of partitions or it can also decrease the no.of partitions
# for example -> df = 4 partitions -> df.repartition(8) -> 8 partitons
# df -> 8 partitions -> df.repartition(4) -> 4 partitions

# coleasce -> it can only be used to decrease the no.of partitions
# for example -> df = 4 partitions -> df.coalesce(2) -> 2 partions

# based on near by partitions, it tries to merge 2 partitions to a single partitions
```

```python
from pyspark.sql import functions as F
```

```python
users_df = spark.table("samples.wanderbricks.users")
bookings_df = spark.table("samples.wanderbricks.bookings")
```

```python
result_df = (
    bookings_df
    .withColumn("parition_id", F.spark_partition_id())
    .groupBy("parition_id")
    .agg(F.count("*").alias("count"))
)

result_df.display()
```

```python
rep_bookings_df = bookings_df.repartition(8)

result_df = (
    rep_bookings_df
    .withColumn("parition_id", F.spark_partition_id())
    .groupBy("parition_id")
    .agg(F.count("*").alias("count"))
)

result_df.display()
```

```python
col_bookings_df = rep_bookings_df.coalesce(4)

result_df = (
    col_bookings_df
    .withColumn("parition_id", F.spark_partition_id())
    .groupBy("parition_id")
    .agg(F.count("*").alias("count"))
)

result_df.display()
```

```python
rep_bookings_df.display()
```

```python
df = rep_bookings_df.repartition(4, "user_id")

result_df = (
    df
    .withColumn("parition_id", F.spark_partition_id())
    .groupBy("parition_id")
    .agg(F.count("*").alias("count"))
)

result_df.display()
```
