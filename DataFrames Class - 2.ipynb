```python
# filter & where

goalkeeprs_df = players_df.filter(F.col("position") == "Goalkeeper")
goalkeeprs_df.display()
```

```python
goalkeeprs_df = (
    players_df
    .filter(
        (F.col("position") == "Goalkeeper") &
        (F.col("foot") == "left")
    )
)

# goalkeeprs_df = players_df.filter((F.col("position") == "Goalkeeper") & (F.col("foot") == "left"))

goalkeeprs_df.display()
```

```python
goalkeeprs_df = (
    players_df
    .where(
        (F.col("position") == "Goalkeeper") &
        (F.col("foot") == "left")
    )
)

goalkeeprs_df.display()
```

```python
df = goalkeeprs_df.drop("first_name", "last_name")
df.display()
```

```python
# date 06-08-2026

df = (
    goalkeeprs_df
    .withColumn(
        "dob_in", # new column name
        F.date_format(F.col("date_of_birth"), "dd-MM-yyyy")  # new-col logic
    )
)
df.display()
```

```python
df = (
    goalkeeprs_df
    .withColumnRenamed(
        "player_id", # old-col name
        "id" # new-col name
    )
)
df.display()
```

```python
df = (
    goalkeeprs_df
    .select(
        F.col("player_id").alias("id"),
        F.col("name").alias("player_name"),
        F.date_format(F.col("date_of_birth"), "dd-MM-yyyy").alias("dob_in")
    )
)
df.display()
```

```python
players_df.createOrReplaceTempView("players")
```

```python
%sql
select * from players;
```

```python
%sql
select  
country_of_citizenship as country, 
count(player_id) as no_of_players 
from players 
group by country_of_citizenship;
```

```python
# find out the country_of_citizenship and no.of players with that citizenship

df = (
    players_df
    .groupBy("country_of_citizenship")
    .count()
)
df.display()
```

```python
# find out the country_of_citizenship and no.of players with that citizenship

df = (
    players_df
    .groupBy("country_of_citizenship")
    .agg(
        F.count("*").alias("no_of_players"), # one-aggregation
        F.round(F.avg("height_in_cm"), 2).alias("avg_height_in_cm"), # secnd-aggregation
    )
)
df.display()
```
