```python
from pyspark.sql import functions as F
```

Let's understand the Spark Execution

```python
# Transformations
trans_df = spark.table("samples.bakehouse.sales_transactions")

result_df = (
    trans_df
    .filter(F.col("paymentMethod") == "mastercard")
    .groupBy("franchiseID")
    .agg(F.sum("totalPrice").alias("total_revenue"))
)

# DAG - Directed Acyclic Graph 
```

```python
# Action  -> catalyst optimizer -> logical plan -> optimized logical plan -> multiple-physical plans -> execution plan (optimized physical plan) -> executes
result_df.show()
```

```python
users_df = spark.table("samples.wanderbricks.users")
bookings_df = spark.table("samples.wanderbricks.bookings")

result_df = (
    users_df.select("user_id", "name")
    .join(
        bookings_df,
        on=["user_id"],
        how="inner"

    )
    .filter(F.year(F.col("check_in")) == 2025)
)

result_df.display()
```

```python
bookings_df = spark.table("samples.wanderbricks.bookings")
agg_df = (
    bookings_df
    .groupBy(F.year(F.col("check_in")).alias("year"))
    .agg(F.count("*").alias("total_records"))
)

agg_df.display()
```

```python
users_df = spark.table("samples.wanderbricks.users")
bookings_df = spark.table("samples.wanderbricks.bookings")

bookings_df = bookings_df.filter(F.year(F.col("check_in")) == 2025) # 57k records

result_df = (
    users_df.select("user_id", "name")
    .join(
        bookings_df,
        on=["user_id"],
        how="inner"
    )
)

result_df.display()
```

```python
# Narrow Transformations --> It won't have any shuffling
# filter, select, limit

# Wide Transformations --> This involes shuffling
# groupBy, join, sort, aggregate,
```
