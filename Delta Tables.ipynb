```python
from pyspark.sql import functions as F
```

```python
emp_df = spark.table("samples.wanderbricks.employees")
emp_df.display()
```

```python
emp_df.count()
```

```python
df = emp_df.groupBy("is_currently_employed").agg(F.count("*").alias("record_count"))
df.display()
```

```python
emp_df.write.format("delta").mode("overwrite").save("abfss://adb-container@quantcloudadls.dfs.core.windows.net/delta_tables/employees")
```

```python
emp_df.write.format("delta").mode("overwrite").saveAsTable("adb_quant_learning.default.employees")
```
